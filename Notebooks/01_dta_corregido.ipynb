## Limpieza con Lógica Geográfica

```python
def limpiar_con_logica_nyc(df, etiqueta):
    antes = len(df)

    # PASO 1: Renombrar fechas primero
    if 'tpep_pickup_datetime' in df.columns:
        df = df.rename(columns={'tpep_pickup_datetime': 'pickup_dt', 'tpep_dropoff_datetime': 'dropoff_dt'})
    else:
        df = df.rename(columns={'lpep_pickup_datetime': 'pickup_dt', 'lpep_dropoff_datetime': 'dropoff_dt'})

    # PASO 2: Calcular duración para poder filtrarla
    df['duration_min'] = (df['dropoff_dt'] - df['pickup_dt']).dt.total_seconds() / 60

    # PASO 3: Aplicar todos los filtros
    df_c = df[
        (df['trip_distance'] > 0.18) &
        (df['trip_distance'] < 100) &
        (df['fare_amount'] > 3.0) &
        (df['PULocationID'] <= 263) &
        (df['DOLocationID'] <= 263) &
        (df['PULocationID'] != df['DOLocationID']) &
        (df['duration_min'] > 1) &
        (df['duration_min'] < 300) &
        (df['fare_amount'] < 600)
    ].copy()

    print(f"{etiqueta}: {antes} → {len(df_c)} registros ({antes - len(df_c)} eliminados)")
    return df_c

# Aplicar a tus 4 dataframes
df_y_mar_clean = limpiar_con_logica_nyc(df_y_marzo, "Yellow Marzo")
df_y_nov_clean = limpiar_con_logica_nyc(df_y_nov, "Yellow Noviembre")
df_g_mar_clean = limpiar_con_logica_nyc(df_g_marzo, "Green Marzo")
df_g_nov_clean = limpiar_con_logica_nyc(df_g_nov, "Green Noviembre")
```

###  LIMPIEZA DE DATOS - TAXI NYC
--------------------------------------

1. trip_distance > 0.1
   Elimina viajes de menos de 180 metros. Un taxi que "se mueve" 50 metros se interpreta como un error
   GPS que no actualizó, no un viaje real.

2. trip_distance < 100
   100 millas = 160 km. Cruzar NYC de punta a punta son ~30 millas. 100 millas
   es imposible dentro de la ciudad (error de GPS o dato corrupto).El area cuadrda otal de la cuidad son aprox 300 millas

3. fare_amount > 3.0
   La tarifa mínima en NYC es $3. Cualquier valor menor es un error de registro
   o un viaje cancelado.

4. PULocationID <= 263 y DOLocationID <= 263
   NYC tiene exactamente 263 zonas oficiales. IDs fuera de rango no existen .

5. PULocationID != DOLocationID
   Origen y destino distintos. Si son iguales, el taxi no se movió entre zonas
   y los ratios precio/km se disparan a infinito.

6. duration_min > 1
   Mínimo 1 minuto de viaje. Viajes de pocos segundos con cobros altos son
   errores de timestamp, no viajes reales.

7. duration_min < 300
   Máximo 5 horas. Un taxi activo 23 horas seguidas es un GPS olvidado
   encendido, no un viaje.

```python
def generar_heatmap_trayectos(df, titulo):
    # 1. Agrupar por ID de recogida y entrega
    # Usamos .size() para contar cuántos viajes hubo en cada combinación
    df_matrix = df.groupby(['PULocationID', 'DOLocationID']).size().reset_index(name='count')

    # 2. Pivotar los datos: Filas (Dropoff), Columnas (Pickup)
    matrix_pivot = df_matrix.pivot(index='DOLocationID', columns='PULocationID', values='count').fillna(0)

    # 3. Graficar
    plt.figure(figsize=(10, 8))
    # Usamos una escala logarítmica (robust=True) para que los aeropuertos no opaquen al resto
    sns.heatmap(matrix_pivot, cmap="YlGnBu", robust=True)
    plt.title(f"Matriz de Trayectos: {titulo}\n(X: Origen | Y: Destino)")
    plt.xlabel("ID Zona Recogida (Pickup)")
    plt.ylabel("ID Zona Entrega (Dropoff)")
    plt.show()

generar_heatmap_trayectos(df_y_mar_clean, "Yellow Taxis - Marzo 2025")
```

## Grafica Sankey rutas

Se mapea la frecuencia de viajes entre cada par de zonas (Origen/Destino). En el Sankey, filtramos el "ruido" para ver solo el movimiento entre las 5 zonas con más demanda.

```python
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "iframe"


df_sankey = df_sample[
    df_sample['PULocationID'].isin(top_5_ids) &
    df_sample['DOLocationID'].isin(top_5_ids)
].copy()

sankey_counts = df_sankey.groupby(['PULocationID', 'DOLocationID']).size().reset_index(name='cantidad')
sankey_counts = sankey_counts[sankey_counts['PULocationID'] != sankey_counts['DOLocationID']]

# Verificación de datos
if len(sankey_counts) > 0:
    labels_left  = [f"Origen: {zone_map[i]}" for i in top_5_ids]
    labels_right = [f"Destino: {zone_map[i]}" for i in top_5_ids]
    labels = labels_left + labels_right

    id_to_left  = {id_real: i   for i, id_real in enumerate(top_5_ids)}
    id_to_right = {id_real: i+5 for i, id_real in enumerate(top_5_ids)}

    sources = sankey_counts['PULocationID'].map(id_to_left).tolist()
    targets = sankey_counts['DOLocationID'].map(id_to_right).tolist()
    values  = sankey_counts['cantidad'].tolist()

    zone_colors = ['#e63946','#2a9d8f','#e9c46a','#457b9d','#f4a261']
    def hex_to_rgba(hex_color, alpha=0.5):
        hex_color = hex_color.lstrip('#')
        r, g, b = tuple(int(hex_color[i:i+2], 16) for i in (0, 2, 4))
        return f'rgba({r},{g},{b},{alpha})'

    link_colors = [hex_to_rgba(zone_colors[s]) for s in sources]
    node_colors = zone_colors + zone_colors

    fig = go.Figure(data=[go.Sankey(
        arrangement='snap',
        node=dict(
            pad=25, thickness=25, label=labels, color=node_colors,
            line=dict(color='white', width=0.5)
        ),
        link=dict(source=sources, target=targets, value=values, color=link_colors)
    )])

    fig.update_layout(
        title_text="Movimiento de Taxis - Top 5 Zonas",
        font_size=12,
        paper_bgcolor='#1e1e2e',
        font_color='white'
    )

    fig.show()
else:
    print("No hay datos suficientes para generar el gráfico.")

```

## Cálculo de Métricas

Se calcula El promedio de duración (minutos) y distancia (millas) de los viajes por cada hora del día, comparando Marzo y Noviembre

```python
def calcular_ratios(df):
    df['duration_min'] = (df['dropoff_dt'] - df['pickup_dt']).dt.total_seconds() / 60
    df['trip_km'] = df['trip_distance'] * 1.609

    # Filtrar registros anómalos ANTES de calcular ratios
    mask = (
        (df['trip_km'] >= 1.0) &          # mínimo 1 km real
        (df['duration_min'] >= 2.0) &     # mínimo 2 minutos
        (df['fare_amount'] > 0)            # tarifa positiva
    )
    df = df[mask].copy()

    df['price_per_km']  = df['fare_amount'] / df['trip_km']
    df['time_per_km']   = df['duration_min'] / df['trip_km']
    df['price_per_min'] = df['fare_amount'] / df['duration_min']
    df['hour']          = df['pickup_dt'].dt.hour

    # CORRECCIÓN: Filtrar outliers de ratios (registros corruptos que pasaron limpieza inicial)
    # En NYC el precio por km raramente supera 25-30 $/km (tarifas aeropuerto incluidas)
    # Un time_per_km > 60 implica más de 1 hora para recorrer 1 km (GPS congelado)
    # Un price_per_min > 20 implica más de 1.200 $/hora (imposible en NYC)
    mask_ratios = (
        (df['price_per_km']  < 30) &
        (df['time_per_km']   < 60) &
        (df['price_per_min'] < 20)
    )
    df = df[mask_ratios].copy()
    print(f"  → Tras filtro de ratios: {len(df)} registros")

    return df

df_yellow = calcular_ratios(df_yellow)
df_green  = calcular_ratios(df_green)
```

### Nota sobre filtros de ratios

**Corrección aplicada:** Se añadieron filtros post-cálculo en `calcular_ratios()` para eliminar registros corruptos que pasaron la limpieza inicial pero generan ratios imposibles:

- `price_per_km < 30` — elimina tarifas >30 $/km (en NYC el máximo real son ~20-25 $/km en aeropuertos)
- `time_per_km < 60` — elimina viajes donde 1 km tardó más de 1 hora (GPS congelado)
- `price_per_min < 20` — elimina tarifas >1.200 $/hora (imposible en taxis regulados)

Estos outliers eran visibles en el Análisis Cualitativo como top trayectos con valores de 316 $/km o 257 min/km.

## "Earnings Per Hour" (EPH)

```python
def analizar_rentabilidad_horaria(df, titulo):
    # Calculamos ingresos por hora de trayecto: (Tarifa / Duración en minutos) * 60
    df['earnings_per_hour'] = (df['fare_amount'] / df['duration_min']) * 60

    # Filtramos valores absurdos (viajes de 10 segundos con tarifa mínima)
    df_plot = df[df['earnings_per_hour'] < 200]

    plt.figure(figsize=(12, 6))
    sns.lineplot(data=df_plot, x='hour', y='earnings_per_hour', hue='month', marker='o')
    plt.axhline(df_plot['earnings_per_hour'].mean(), color='red', linestyle='--', label='Media Global')
    plt.title(f"Rentabilidad Horaria ({titulo}): $/Hora efectiva de viaje")
    plt.ylabel("Dólares por Hora ($/h)")
    plt.xlabel("Hora del Día")
    plt.legend()
    plt.show()

analizar_rentabilidad_horaria(df_yellow, "Yellow Taxi")
analizar_rentabilidad_horaria(df_green, "Green Taxi")
```
